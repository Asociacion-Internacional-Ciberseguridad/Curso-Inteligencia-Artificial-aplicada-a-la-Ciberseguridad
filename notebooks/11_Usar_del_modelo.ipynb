```python
import pandas as pd
import joblib
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score

# 1. Cargar el dataset sintético
df = pd.read_csv('iris_sintetico_con_etiqueta.csv')

# 2. Cargar el modelo entrenado
modelo_cargado = joblib.load('modelo_iris.pkl')

# Cargar metadatos de iris para nombres de especies
iris = load_iris()

# 3. Preparar los datos del CSV
# Eliminamos las columnas 'etiqueta' y 'species' para tener solo las características (X)
X_test = df.drop(columns=['etiqueta', 'species'])
# Guardamos las etiquetas reales para comparar (y)
y_real = df['etiqueta']

# Asegurar que las columnas tengan los nombres que el modelo espera
X_test.columns = iris.feature_names

# 4. Realizar la predicción masiva
predicciones = modelo_cargado.predict(X_test)

# 5. Mostrar resultados
# Añadimos las predicciones al DataFrame original para visualizar mejor
df['prediccion_modelo'] = predicciones
df['nombre_especie_predicha'] = [iris.target_names[p] for p in predicciones]

print("Primeras filas con las predicciones:")
print(df[['etiqueta', 'prediccion_modelo', 'nombre_especie_predicha']].head())

# 6. Calcular el rendimiento
precision = accuracy_score(y_real, predicciones)
print(f'\nPrecisión del modelo con el archivo CSV: {precision * 100:.2f}%')

# Opcional: Guardar los resultados en un nuevo archivo
# df.to_csv('resultados_prediccion.csv', index=False)
```
